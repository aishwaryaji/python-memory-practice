Q1 Boston Housing Price (Regression – DNN)

```python
import tensorflow as tf
from tensorflow import keras
from keras import layers

# Load data
(x_train, y_train), (x_test, y_test) = keras.datasets.boston_housing.load_data()

# Normalize
mean = x_train.mean(axis=0)
std = x_train.std(axis=0)
x_train = (x_train - mean) / std
x_test = (x_test - mean) / std

# Model
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(13,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# Train
model.fit(x_train, y_train, epochs=50, batch_size=16, verbose=0)

# Evaluate
loss, mae = model.evaluate(x_test, y_test)
print("MAE:", mae)
```

MAE ≈ 2–4 → good model (price error in $1000s)

Q2 IMDB Sentiment Classification

```python
from tensorflow import keras
from keras import layers

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=10000)

# Pad sequences
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=200)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=200)

model = keras.Sequential([
    layers.Embedding(10000, 128, input_length=200),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=3, batch_size=32)

model.evaluate(x_test, y_test)
```

Binary classification (positive/negative)

Q3 Reuters News Classification

```python
from tensorflow import keras
from keras import layers

(x_train, y_train), (x_test, y_test) = keras.datasets.reuters.load_data(num_words=10000)

x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=100)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=100)

model = keras.Sequential([
    layers.Embedding(10000, 128),
    layers.GlobalAveragePooling1D(),
    layers.Dense(64, activation='relu'),
    layers.Dense(46, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=5, batch_size=32)

model.evaluate(x_test, y_test)
```

 46 classes → accuracy lower than IMDB is normal

Q4. CNN MNIST

```python
from tensorflow import keras
from keras import layers

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

x_train = x_train.reshape(-1,28,28,1)/255.0
x_test = x_test.reshape(-1,28,28,1)/255.0

model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=3)

model.evaluate(x_test, y_test)
```

Very high accuracy expected

Q5. Pretrained Model (MobileNetV2)

```python
from tensorflow import keras
from keras import layers

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

x_train = x_train.reshape(-1,28,28,1)
x_test = x_test.reshape(-1,28,28,1)

# Convert to 3 channels
x_train = tf.image.resize(x_train, (32,32)) / 255.0
x_test = tf.image.resize(x_test, (32,32)) / 255.0

x_train = tf.image.grayscale_to_rgb(x_train)
x_test = tf.image.grayscale_to_rgb(x_test)

base = keras.applications.MobileNetV2(input_shape=(32,32,3), include_top=False, weights='imagenet')
base.trainable = False

model = keras.Sequential([
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=3)

model.evaluate(x_test, y_test)
```

Slightly lower than CNN due to resizing

Q6. LSTM Time Series

```python
import numpy as np
from tensorflow import keras
from keras import layers

# Dummy data
data = np.sin(np.arange(0,100,0.1))

# Prepare sequences
X, y = [], []
for i in range(50, len(data)):
    X.append(data[i-50:i])
    y.append(data[i])

X, y = np.array(X), np.array(y)
X = X.reshape((X.shape[0], X.shape[1], 1))

model = keras.Sequential([
    layers.LSTM(50, input_shape=(50,1)),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=5)
```

Loss keeps decreasing

Q7. Autoencoder

```python
from tensorflow import keras
from keras import layers

(x_train, _), (x_test, _) = keras.datasets.mnist.load_data()

x_train = x_train.reshape(-1,784)/255.0
x_test = x_test.reshape(-1,784)/255.0

# Encoder
input_img = keras.Input(shape=(784,))
encoded = layers.Dense(64, activation='relu')(input_img)

# Decoder
decoded = layers.Dense(784, activation='sigmoid')(encoded)

autoencoder = keras.Model(input_img, decoded)

autoencoder.compile(optimizer='adam', loss='binary_crossentropy')
autoencoder.fit(x_train, x_train, epochs=5)

autoencoder.evaluate(x_test, x_test)
```

Lower loss = better reconstruction
